QUESTAO 2

```
notas = [4.5, 6.0, 7.8, 9.0, 5.5, 8.2]

classificacoes = []

aprovados = 0
reprovados = 0

for nota in notas:
    if nota < 5:
        classificacoes.append("Reprovado")
        reprovados += 1
    elif 5 <= nota < 7:
        classificacoes.append("Recuperação")
    else:
        classificacoes.append("Aprovado")
        aprovados += 1

percentual_reprovados = (reprovados / len(notas)) * 100

print("Classificações:", classificacoes)
print("Quantidade de aprovados:", aprovados)
print("Percentual de reprovados:", percentual_reprovados, "%")
```

QUESTAO 3

```
consumo = [120, 95, 200, 180, 75, 220, 160]

consumo_alto = []
total = 0

for c in consumo:
    total += c

    if c > 180:
        consumo_alto.append(c)

media = total / len(consumo)

print("Consumos altos:", consumo_alto)
print("Total da semana:", total)
print("Média semanal:", media)

if len(consumo_alto) > 2:
    print("ALERTA: Alto consumo em muitos dias!")
```

QUESTAO 4

```
precos = [50, 120, 30, 200, 80, 15]

precos_finais = []
economia_total = 0

for preco in precos:
    if preco < 50:
        desconto = preco * 0.05
    elif 50 <= preco < 150:
        desconto = preco * 0.10
    else:
        desconto = preco * 0.15

    preco_final = preco - desconto

    precos_finais.append(preco_final)
    economia_total += desconto

total_compra = sum(precos_finais)

print("Preços finais:", precos_finais)
print("Total com desconto:", total_compra)
print("Economia total:", economia_total)
```
